```python
# reading a PDF file

from langchain_community.document_loaders import PyPDFLoader
pdf_loader = PyPDFLoader("attention.pdf")
pdf_documents = pdf_loader.load() # create the pdf document from the file
pdf_documents[0].page_content[:500] # print the first 500 characters of the pdf document
```

## Recursive Text Splitter

A recursive text splitter divides large documents into smaller chunks while preserving meaningful context.

It tries these default separators in order:

1. Paragraph breaks (`\n\n`)
2. Line breaks (`\n`)
3. Spaces (` `)
4. Individual characters (`""` — an empty separator)

If a chunk is still larger than `chunk_size`, the splitter uses the next separator to divide it further.

- `chunk_size`: Maximum chunk size, measured in characters by default.
- `chunk_overlap`: Target overlap between consecutive chunks. Actual overlap may be smaller.

Overlap helps preserve context across chunk boundaries.

```python
# Recursive character text splitter
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)

documents = text_splitter.split_documents(pdf_documents) # split the pdf document into chunks

documents[0].page_content[:500] # print the first 500 characters of the first chunk
```

```python
print(documents[0]) # print the first document in the final_documents list
print(documents[0].metadata) # print the metadata of the first document in the final_documents list
```

## Character Text Splitter

A character text splitter divides a document into smaller chunks using a single separator, such as a paragraph break (`\n\n`). It combines the resulting pieces into chunks based on a target size.

- **`separator`**: Where to split the text.
- **`chunk_size`**: Target maximum size, measured in characters by default.
- **`chunk_overlap`**: Target amount of text shared between consecutive chunks to preserve context.

If a piece exceeds `chunk_size`, it may remain oversized because the splitter does not automatically try smaller separators.

### Difference from Recursive Text Splitter

- **CharacterTextSplitter** uses one separator.
- **RecursiveCharacterTextSplitter** tries a sequence of separators to split oversized pieces further.

```python
from langchain_community.document_loaders import TextLoader

loader = TextLoader("speech.txt")
documents = loader.load()
documents[0].page_content[:500] # print the first 500 characters of the text document
```

```python
#text splitter
from langchain_text_splitters import CharacterTextSplitter
text_splitter = CharacterTextSplitter(separator="\n\n", chunk_size=100, chunk_overlap=20)
text_splitter.split_documents(documents) # split the text document into chunks
```

```python
# read with file reader

text = ""
with open("speech.txt", "r") as f:
    text = f.read() 

text_splitter = CharacterTextSplitter(separator="\n\n", chunk_size=100, chunk_overlap=20)
text = text_splitter.create_documents([text]) # split the text into chunks
text[0]
text[1]
```
