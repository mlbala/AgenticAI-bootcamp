## HuggingFace Sentence Transformers

**Sentence Transformers** (the `sentence-transformers` library, built on top of HuggingFace's `transformers`) are models specifically fine-tuned to produce high-quality sentence/paragraph embeddings — unlike raw BERT-style models, they're trained (often with contrastive/siamese objectives) so that semantically similar sentences land close together in vector space.

**Why use them here:**
- **Free & local** — the model runs on your own machine (CPU/GPU), no API key or per-call cost like OpenAI embeddings.
- **Open-source models** — e.g., `all-MiniLM-L6-v2`, `all-mpnet-base-v2` — hosted on the HuggingFace Hub.
- In LangChain, we access these via `HuggingFaceEmbeddings` (or `HuggingFaceInferenceAPIEmbeddings` for hosted inference), which wraps `sentence-transformers` under the hood and plugs directly into the same embedding/vector-store pipelines used for OpenAI embeddings.



```python
import os
from dotenv import load_dotenv
# Change working directory to the parent folder
os.chdir(os.path.abspath(".."))
load_dotenv()
```

```python
os.environ['HF_TOKEN'] = os.getenv("HF_KEY")  # Set the Hugging Face token as an environment variable
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")

documents = [
    "This is the first document.",
    "This is the second document.",
    "This is the third document about LangChain and embeddings.",
]

doc_embeddings = embeddings.embed_documents(documents)
print(f"Number of documents embedded: {len(doc_embeddings)}")
print(f"Embedding dimension: {len(doc_embeddings[0])}")
```
