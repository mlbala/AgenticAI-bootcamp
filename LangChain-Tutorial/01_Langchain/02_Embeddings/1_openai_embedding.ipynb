## Embeddings & OpenAI Embedding Models

**What is an embedding?**
An embedding is a technique that converts text (a word, sentence, or document) into a fixed-length vector of numbers (floating-point values) such that the vector captures the *semantic meaning* of the text. Texts with similar meaning end up with vectors that are close together in this high-dimensional vector space, while unrelated texts end up far apart. This lets us do things like semantic search, clustering, recommendation, and retrieval-augmented generation (RAG) using simple vector math (e.g., cosine similarity) instead of exact keyword matching.

**How embedding models work**
An embedding model is typically a neural network (often transformer-based) trained so that pieces of text with similar context/meaning are mapped near each other in vector space. Once trained, the model takes any input text and outputs a dense vector (e.g., 1536 dimensions) representing that text.

**Technique we're using here: OpenAI Embeddings**
In this notebook, we use OpenAI's embedding models (via the `langchain_openai` / `openai` SDK) to generate embeddings for text. Key points:

- We call OpenAI's embeddings API (e.g., `text-embedding-3-small` or `text-embedding-3-large`) which returns a vector representation for each input text.
- These models are pre-trained by OpenAI on massive text corpora — we don't train anything ourselves, we simply send text and receive back its vector representation.
- The resulting embeddings can be stored in a vector database (like FAISS, Chroma, Pinecone, etc.) and later queried using similarity search — this is the foundation of RAG pipelines.
- Authentication is handled via the `OPENAI_API_KEY` environment variable, which we load using `python-dotenv` in the cells below.



```python
import os
from dotenv import load_dotenv
# Change working directory to the parent folder
os.chdir(os.path.abspath(".."))
load_dotenv()
```

```python
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
```

```python
from langchain_openai import OpenAIEmbeddings
embeddings = OpenAIEmbeddings(model="text-embedding-3-large")
embeddings
```

```python
text="This is tutorial for Openai Embeddings. It is a great tutorial for openai embeddings using  LangChain."

# convert the text to vector representation
query_result = embeddings.embed_query(text)
query_result
```

```python
# restrict to 1024 
embeddings = OpenAIEmbeddings(model="text-embedding-3-large", dimensions=1024)
text="This is tutorial for Openai Embeddings. It is a great tutorial for openai embeddings using  LangChain."

# convert the text to vector representation
query_result = embeddings.embed_query(text)
query_result
```

```python
len(query_result)
```

```python
from langchain_community.document_loaders import TextLoader
loader = TextLoader("speech.txt")
docs = loader.load()
docs
```

```python
# Recursive character text splitter
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(chunk_size=100, chunk_overlap=50)

documents = text_splitter.split_documents(docs) # split the  document into chunks

documents[0].page_content[:500] # print the first 500 characters of the first chunk
```

```python
# store in vector database
from langchain_chroma import Chroma
vectorstore = Chroma.from_documents(documents, embeddings, collection_name="speech", persist_directory="db")
```

```python
query = "What is democracy?"
results = vectorstore.similarity_search(query, k=2)

print("Results:")
print(results[0].page_content)
```
